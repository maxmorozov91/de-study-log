```python
import os
from contextlib import closing
from pathlib import Path

import polars as pl
import psycopg2
from dotenv import load_dotenv

load_dotenv()

dsn = (
  f"host=localhost dbname={os.environ["EXAMPLE_DB"]} "
  f"user={os.environ["POSTGRES_USER"]} "
  f"password={os.environ["POSTGRES_PASSWORD"]}"
)

dsn_salesdb = (
    f"host=127.0.0.1 port=5432 dbname={os.environ["SALES_DB"]} "
    f"user={os.environ["POSTGRES_USER"]} "
    f"password={os.environ["POSTGRES_PASSWORD"]}"
)

def run_problem(prob_num: str) -> pl.DataFrame:
    problem_dpath = Path.cwd() / "sql" / "leetcode" / prob_num
    setup_script = problem_dpath / "setup.sql"
    solution_script = problem_dpath / "solution.sql"
    dsn = f"""
host=127.0.0.1 dbname={os.environ["LEETCODE_DB"]}
user={os.environ["POSTGRES_USER"]} password={os.environ["POSTGRES_PASSWORD"]}
"""
    with closing(psycopg2.connect(dsn)) as conn:
        with conn, conn.cursor() as cur:
            cur.execute(setup_script.read_text())
            print(f"data for {prob_num} prepared:", cur.rowcount)
        
        with conn:
            return pl.read_database(solution_script.read_text(), conn)
```

## String Functions

```python
sql = """
SELECT
    CONCAT_WS(' ', LTRIM(first_name), country)
FROM customers
;
"""

with closing(psycopg2.connect(dsn)) as conn:
    df = pl.read_database(sql, conn)

print(df)
```

```python
sql = """
SELECT
    UPPER(first_name)
  , LOWER(first_name)
FROM customers
;
"""

with closing(psycopg2.connect(dsn)) as conn:
    df = pl.read_database(sql, conn)

print(df)
```

```python
sql = """
SELECT
    *
FROM customers
WHERE first_name != TRIM(first_name)
;
"""

with closing(psycopg2.connect(dsn)) as conn:
    df = pl.read_database(sql, conn)

print(df)
```

```python
sql = """
SELECT REPLACE('123-456-7890', '-', '')
"""

with closing(psycopg2.connect(dsn)) as conn:
    with conn, conn.cursor() as cur:
        cur.execute(sql)
        print(cur.fetchall())
```

```python
sql = """
SELECT
    BTRIM(first_name)
  , LENGTH(BTRIM(first_name))
FROM customers
"""

with closing(psycopg2.connect(dsn)) as conn:
    df = pl.read_database(sql, conn)

df
```

```python
sql = """
SELECT
    LEFT(LTRIM(first_name), 2)
  , RIGHT(RTRIM(first_name), 2)
FROM customers
"""

with closing(psycopg2.connect(dsn)) as conn:
    df = pl.read_database(sql, conn)

df
```

```python
sql = """
SELECT
    SUBSTRING(LTRIM(first_name), 2, LENGTH(first_name))
FROM customers
"""

with closing(psycopg2.connect(dsn)) as conn:
    df = pl.read_database(sql, conn)

df
```

```python
sql = """
SELECT
    2.516           AS orig_num
  , ROUND(2.516, 2) AS round_2
  , ROUND(2.516, 1) AS round_1
  , ROUND(2.516)    AS round
"""

with closing(psycopg2.connect(dsn)) as conn:
    df = pl.read_database(sql, conn)

df
```

## Date & Time Functions

```python
sql = """
SELECT
    order_date
  , EXTRACT(YEAR   FROM order_date) AS year
  , EXTRACT(MONTH  FROM order_date) AS month
  , EXTRACT(WEEK   FROM order_date) AS week
  , EXTRACT(ISODOW FROM order_date) AS day_of_week
FROM orders
"""

with closing(psycopg2.connect(dsn)) as conn:
    df = pl.read_database(sql, conn)

print(df)
```

```python
sql = """
SELECT
    order_date
  , DATE_TRUNC('week', order_date) AS trunc_week
  , DATE_TRUNC('month', order_date) AS trunc_month
  , DATE_TRUNC('year', order_date) AS trunc_year
FROM orders
"""

with closing(psycopg2.connect(dsn)) as conn:
    df = pl.read_database(sql, conn)

print(df)
```

```python
sql = """
SELECT
    order_date
  , TO_CHAR(order_date, 'YYYY-MM') AS year_month
  , TO_CHAR(order_date, 'Month') AS month
  , TO_CHAR(order_date, 'FMMonth') AS fm_month
  , TO_CHAR(order_date, 'Day') AS day
  , TO_CHAR(order_date, 'Dy') AS short_day
FROM orders
"""

with closing(psycopg2.connect(dsn)) as conn:
    df = pl.read_database(sql, conn)

print(df)
```
